```python
# local scope
def f():
    x = 10
    print(x)
f()
```

```python
# Global scope
x = 20
def f2():
    print(x)
f2()
print(x)
```

```python
# enclosing scope
def outer():
    x = "outer"
   # print(x)
    def inner():
        y = "inner"
        print(x,y)
    inner()
outer()
print(x)
```

```python
# Built in scope
def f3():
    l = [1,2,3,4]
    return sum(l)
f3()
```

```python
# Global scope
x = 20
def f2():
    global x
    x =x+10
    print(x)
f2()
print(x)
```

```python
# Immutable 
def f(x):
    x = 20
    print(id(x))
    print(x)
a=10
print(id(a))
f(a)
```

```python
#Mutabletype
def f(x):
    x.append(5)
    print(id(x))
    print(x)
a=[1,2,3,4,]
print(id(a))
f(a)
```

```python
#mutabletype reassign the addres will changes
def f(x):
    x.append(10)
    print(id(x))
    x = [1,2,3,4,10]
    print(id(x))
    print(x)
a=[1,2,3,4,]
print(id(a))
f(a)
```
