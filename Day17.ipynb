```python
# List Comprehension
lst = [1, 2, 3, 4, 5]

result = [x**3 for x in lst]

print(result)
```

```python
[x for x in lst if x % 2 != 0]
```

```python
M = [[1,2,3],[4,5,6],[7,8,9]]
r =[(row,column) for row in range(len(M)) for column in range(len(M[row]))]
print(r)
```

```python
M = [[1,2,3],[4,5,6],[7,8,9]]
r =[M[row][column] for row in range(len(M))  for column in range(len(M[row]))]
print(r)
```

```python
# Generators
def generator():
    yield 1
    yield 2
    yield 3
r = generator()
print(next(r))
print(next(r))
print(next(r))
```

```python
def generator():
    yield 1
    yield 2
    yield 3
r = generator()
for i in r:
    print(i)
```

```python
def generator(x):
    x = x+1
    yield x
    x+=10
    yield x
    x+=20
    yield x
r = generator(10)
for i in r:
    print(i)
```
