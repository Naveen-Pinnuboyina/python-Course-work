```python
def factorial(n):
    if n == 1 or n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
def sum_of_natural_numbers(n):
    if n==2:
        return 2
    return n+sum_of_natural_numbers(n-2)
print(sum_of_natural_numbers(10))
```

```python
def sum_of_natural_numbers(n):
    if n%2==0:
        if n==2:
            return 2
        return n+sum_of_natural_numbers(n-2)
    else:
        return sum_of_natural_numbers(n+1)
print(sum_of_natural_numbers(10))
```

```python
def pnc(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count +=1
    if count==2:
        return True
    else:
        return False
pnc(11)
```

```python
def pnc(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count +=1
    if count==2:
        return True
    else:
        return False
num =100
def pnc_list(num):
    l =[]
    for number in range(2,num+1):
        if pnc(number):
            l.append(number)
    return l
num = int(input())
pnc_list(num)
```
