```python
#collections
import collections
from collections import Counter
name = "Naveen"
r = Counter(name)
print(r)
nums = [1,2,3,4,5,6]
n = Counter(nums)
print(n)
```

```python
from collections import defaultdict
d = defaultdict(int)
d["ball"] = 1
print(d["name"])
```

```python
from collections import deque
d = deque([1,2,3])
d.append(4)
d.appendleft(10)
d.pop()
d.popleft()
print(d)
```

```python
# itertools
from itertools import permutations
numbers = [1,2,3,4]
print(list(permutations(numbers)))
```

```python
from itertools import combinations
numbers = [1,2,3,4]
print(list(combinations(numbers,2)))
```

```python
from itertools import product
a = [1,2,4]
b =['A','D','E']
print(list(product(a,b)))
```
