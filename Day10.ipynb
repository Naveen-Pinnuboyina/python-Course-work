```python
seat = input("Enter Your Seat Type:")
days = int(input("Enter your Booking Days: "))
festival = input("Enter Festival or Not: ").lower() == "true"
age = int(input("Enter Your Age: "))

price = 5000

if seat == "Business":
    price *= 1.4
elif seat == "premium":
    price *= 1.2

if days >= 30:
    price *= 0.9
elif days <= 7:
    price *= 1.25

if festival:
    price *= 1.2

if age > 60:
    price *= 0.85
print(price)

```
