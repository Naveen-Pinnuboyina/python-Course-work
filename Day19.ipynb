````python
import random
characters = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
characters += "abcdefghijklmnopqrstuvwxyz"
characters += "0123456789"
characters += "@#$%^&*"
length = int(input())
password = "" 
for i in range(length):
    ch = random.choice(characters)
    password += ch
print(password)
````

````python
UPPER_CASE = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
lower_case = "abcdefghijklmnopqrstuvwxyz"
digits = "0123456789"
special_characters = "@!#$%^&*"
password = []
length =int(input())
if length < 4:
    print(f"your password lentgh is {length}, choose password length higher")
else:
    password.append(random.choice(UPPER_CASE))
    password.append(random.choice(lower_case))
    password.append(random.choice(digits))
    password.append(random.choice(special_characters))
for i in range(length-4):
    all_characters = UPPER_CASE + lower_case + digits + special_characters
    ch = random.choice(all_characters)
    password.append(ch) 
    random.shuffle(password)
new_password = "".join(password)
print(new_password)
````

````python
import time
from IPython.display import clear_output
seconds = 0
minutes = 0
hours = 0
while True:
    print(f"{hours}:{minutes}:{seconds}")
    time.sleep(1)
    seconds+=1
    if seconds == 60:
        minutes+=1
        seconds = 0
        if minutes == 1:
            print("Tick tick ......1 minute Alarm completed")
            break
            hours += 1
            minutes = 0
    clear_output(wait = True)
        
````

````python
```python
import random

rolls = 0
count = 0

while True:
    input("Please click Enter to throw the dice: ")

    # Throw two dice
    dice1 = random.randint(1, 6)
    dice2 = random.randint(1, 6)

    # Add both dice values
    total = dice1 + dice2

    if count + total <= 100:
        count += total

    rolls += 1

    print(f"Dice 1: {dice1}")
    print(f"Dice 2: {dice2}")
    print(f"Total: {total}")
    print(f"You are at {count} on {rolls} rolls")

    if count == 100:
        print(f"You won the match on {rolls} rolls!")
        break
```
````
